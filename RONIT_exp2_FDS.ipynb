```python
from google.colab import files
import pandas as pd
import numpy as np

uploaded = files.upload()

file_name = next(iter(uploaded))
df = pd.read_csv(file_name)

print("Dataset loaded successfully!")
print("File name:", file_name)
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])

display(df.head())
```

```python
print("Activity 1")
print("Information and Description of Dataset")

print("\nDataset Information:")
df.info()

print("\nDataset Shape:")
print(df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nStatistical Description:")
display(df.describe())

print("\nFirst 5 Records:")
display(df.head())

print("\nLast 5 Records:")
display(df.tail())
```

```python
print("Activity 2")
print("Checking Null Values")

print("\nNull values in each column:")
print(df.isnull().sum())

print("\nTotal null values:")
print(df.isnull().sum().sum())

print("\nColumns containing null values:")
print(df.columns[df.isnull().any()].tolist())
```

```python
print("Activity 3")
print("Handling Missing Values")

df.fillna(0, inplace=True)

print("\nNull values after handling:")
print(df.isnull().sum())

print("\nTotal null values after handling:")
print(df.isnull().sum().sum())

print("\nDataset shape:")
print(df.shape)
```

```python
print("Activity 4")
print("Sorting and Displaying Top Records")

print("\nTop 5 EVs by Range:")
display(
    df.sort_values(by='range_km', ascending=False)
    .head(5)
)

print("\nTop 5 EVs by Price:")
display(
    df.sort_values(by='price_usd', ascending=False)
    .head(5)
)

print("\nTop 5 EVs by Battery Capacity:")
display(
    df.sort_values(by='battery_kwh', ascending=False)
    .head(5)
)

print("\nTop 5 EVs by Efficiency Score:")
display(
    df.sort_values(by='efficiency_score', ascending=False)
    .head(5)
)
```

```python
print("Activity 5")
print("Frequency Listing")

print("\nFrequency of Manufacturers:")
print(df['manufacturer'].value_counts())

print("\nFrequency of Release Years:")
print(df['release_year'].value_counts().sort_index())
```

```python
print("Activity 6")
print("Indexing and Selecting Data")

print("\nFirst 10 rows:")
display(df.head(10))

print("\nSelected columns:")
display(df[['manufacturer', 'model', 'battery_kwh', 'range_km', 'price_usd']].head(10))

print("\nRows from index 10 to 20:")
display(df.iloc[10:21])

print("\nEVs with range greater than 500 km:")
display(df.loc[df['range_km'] > 500].head(10))
```
