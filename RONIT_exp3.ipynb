```python
import math



def calculate_mean(data):
    total = 0
    count = 0

    for value in data:
        total = total + value
        count = count + 1

    return total / count




def calculate_median(data):
    sorted_data = sorted(data)
    n = len(sorted_data)

    if n % 2 == 0:
        middle1 = sorted_data[(n // 2) - 1]
        middle2 = sorted_data[n // 2]
        median = (middle1 + middle2) / 2
    else:
        median = sorted_data[n // 2]

    return median




def calculate_mode(data):
    frequency = {}

    for value in data:
        if value in frequency:
            frequency[value] = frequency[value] + 1
        else:
            frequency[value] = 1

    highest_frequency = 0

    for value in frequency:
        if frequency[value] > highest_frequency:
            highest_frequency = frequency[value]

    modes = []

    for value in frequency:
        if frequency[value] == highest_frequency:
            modes.append(value)

    return modes




def calculate_variance(data):
    mean = calculate_mean(data)

    total = 0

    for value in data:
        difference = value - mean
        squared_difference = difference ** 2
        total = total + squared_difference

    variance = total / len(data)

    return variance




def calculate_standard_deviation(data):
    variance = calculate_variance(data)
    return math.sqrt(variance)




def calculate_iqr(data):
    sorted_data = sorted(data)

    n = len(sorted_data)

    if n % 2 == 0:
        lower_half = sorted_data[:n // 2]
        upper_half = sorted_data[n // 2:]
    else:
        lower_half = sorted_data[:n // 2]
        upper_half = sorted_data[(n // 2) + 1:]

    Q1 = calculate_median(lower_half)
    Q3 = calculate_median(upper_half)

    IQR = Q3 - Q1

    return Q1, Q3, IQR




range_data = []
price_data = []
battery_data = []
acceleration_data = []
efficiency_data = []
cargo_data = []


for value in df['range_km']:
    if not pd.isna(value):
        range_data.append(value)


for value in df['price_usd']:
    if not pd.isna(value):
        price_data.append(value)


for value in df['battery_kwh']:
    if not pd.isna(value):
        battery_data.append(value)


for value in df['acceleration_0_100_kmph']:
    if not pd.isna(value):
        acceleration_data.append(value)


for value in df['efficiency_score']:
    if not pd.isna(value):
        efficiency_data.append(value)


for value in df['cargo_space_liters']:
    if not pd.isna(value):
        cargo_data.append(value)




mean_range = calculate_mean(range_data)

print("1. Mean ")
print("Attribute: range_km")
print("Mean Range =", round(mean_range, 2), "km")



median_price = calculate_median(price_data)

print(" 2. Median ")
print("Attribute: price_usd")
print("Median Price =", round(median_price, 2), "USD")



mode_battery = calculate_mode(battery_data)

print(" 3. Mode ")
print("Attribute: battery_kwh")
print("Mode Battery Capacity =", mode_battery, "kWh")


variance_acceleration = calculate_variance(acceleration_data)

print("4. Variance ")
print("Attribute: acceleration_0_100_kmph")
print("Variance =", round(variance_acceleration, 4))



standard_deviation = calculate_standard_deviation(efficiency_data)

print("5. Standard Deviation ")
print("Attribute: efficiency_score")
print("Standard Deviation =", round(standard_deviation, 4))



Q1, Q3, IQR = calculate_iqr(cargo_data)

print("6. Interquartile Range ")
print("Attribute: cargo_space_liters")
print("Q1 =", round(Q1, 2))
print("Q3 =", round(Q3, 2))
print("IQR =", round(IQR, 2))





print("ACTIVITY 2 FINAL RESULTS")


print("Mean               =", round(mean_range, 2))
print("Median             =", round(median_price, 2))
print("Mode               =", mode_battery)
print("Variance            =", round(variance_acceleration, 4))
print("Standard Deviation  =", round(standard_deviation, 4))
print("Q1                  =", round(Q1, 2))
print("Q3                  =", round(Q3, 2))
print("IQR                 =", round(IQR, 2))

```

```python


import pandas as pd
import math


data = df['range_km'].dropna()

print("ACTIVITY 3")
print("Grouped Data Analysis")
print("Attribute Used: range_km")
print("Number of observations:", len(data))



bins = [0, 100, 200, 300, 400, 500, 600, 700, 800]

labels = [
    '0-100',
    '101-200',
    '201-300',
    '301-400',
    '401-500',
    '501-600',
    '601-700',
    '701-800'
]


groups = pd.cut(
    data,
    bins=bins,
    labels=labels,
    include_lowest=True
)


frequency = groups.value_counts().sort_index()


midpoints = [
    50,
    150.5,
    250.5,
    350.5,
    450.5,
    550.5,
    650.5,
    750.5
]


grouped_data = pd.DataFrame({
    'Class Interval': labels,
    'Frequency': frequency.values,
    'Midpoint': midpoints
})


N = grouped_data['Frequency'].sum()


grouped_data['f × x'] = (
    grouped_data['Frequency'] *
    grouped_data['Midpoint']
)


grouped_data['Cumulative Frequency'] = (
    grouped_data['Frequency'].cumsum()
)

print("GROUPED DATA TABLE ")
display(grouped_data)

print("Total Frequency =", N)



sum_fx = grouped_data['f × x'].sum()

grouped_mean = sum_fx / N

print("\n========== GROUPED MEAN ==========")
print("Σ(f × x) =", sum_fx)
print("Σf =", N)
print("Grouped Mean =", round(grouped_mean, 2), "km")



median_position = N / 2

median_class_index = (
    grouped_data['Cumulative Frequency']
    >= median_position
).idxmax()

median_class = grouped_data.iloc[median_class_index]

L = bins[median_class_index]
f = median_class['Frequency']

if median_class_index == 0:
    CF = 0
else:
    CF = grouped_data.iloc[
        median_class_index - 1
    ]['Cumulative Frequency']

h = 100

grouped_median = L + (
    (median_position - CF) / f
) * h

print("GROUPED MEDIAN ")
print("N/2 =", median_position)
print("Median Class =", median_class['Class Interval'])
print("Grouped Median =", round(grouped_median, 2), "km")



modal_class_index = grouped_data['Frequency'].idxmax()

modal_class = grouped_data.iloc[modal_class_index]

fm = modal_class['Frequency']

if modal_class_index == 0:
    f1 = 0
else:
    f1 = grouped_data.iloc[
        modal_class_index - 1
    ]['Frequency']

if modal_class_index == len(grouped_data) - 1:
    f2 = 0
else:
    f2 = grouped_data.iloc[
        modal_class_index + 1
    ]['Frequency']

L = bins[modal_class_index]

grouped_mode = L + (
    (fm - f1) /
    ((fm - f1) + (fm - f2))
) * h

print(" GROUPED MODE")
print("Modal Class =", modal_class['Class Interval'])
print("Grouped Mode =", round(grouped_mode, 2), "km")



grouped_data['f × x²'] = (
    grouped_data['Frequency'] *
    grouped_data['Midpoint'] ** 2
)

sum_fx2 = grouped_data['f × x²'].sum()

grouped_variance = (
    sum_fx2 / N
) - (grouped_mean ** 2)

print(" GROUPED VARIANCE")
print("Σ(f × x²) =", sum_fx2)
print("Grouped Variance =", round(grouped_variance, 2))



grouped_std = math.sqrt(grouped_variance)

print("\n========== GROUPED STANDARD DEVIATION ==========")
print(
    "Grouped Standard Deviation =",
    round(grouped_std, 2),
    "km"
)



Q1_position = N / 4

Q1_class_index = (
    grouped_data['Cumulative Frequency']
    >= Q1_position
).idxmax()

Q1_class = grouped_data.iloc[Q1_class_index]

L1 = bins[Q1_class_index]
f_Q1 = Q1_class['Frequency']

if Q1_class_index == 0:
    CF_Q1 = 0
else:
    CF_Q1 = grouped_data.iloc[
        Q1_class_index - 1
    ]['Cumulative Frequency']

Q1 = L1 + (
    (Q1_position - CF_Q1) / f_Q1
) * h



Q3_position = 3 * N / 4

Q3_class_index = (
    grouped_data['Cumulative Frequency']
    >= Q3_position
).idxmax()

Q3_class = grouped_data.iloc[Q3_class_index]

L3 = bins[Q3_class_index]
f_Q3 = Q3_class['Frequency']

if Q3_class_index == 0:
    CF_Q3 = 0
else:
    CF_Q3 = grouped_data.iloc[
        Q3_class_index - 1
    ]['Cumulative Frequency']

Q3 = L3 + (
    (Q3_position - CF_Q3) / f_Q3
) * h



grouped_IQR = Q3 - Q1

print(" GROUPED IQR")
print("Q1 =", round(Q1, 2), "km")
print("Q3 =", round(Q3, 2), "km")
print("Grouped IQR =", round(grouped_IQR, 2), "km")



print("       ACTIVITY 3 FINAL RESULTS")


print("Attribute              : range_km")
print("Total Frequency         :", N)
print("Grouped Mean            :", round(grouped_mean, 2), "km")
print("Grouped Median          :", round(grouped_median, 2), "km")
print("Grouped Mode            :", round(grouped_mode, 2), "km")
print("Grouped Variance        :", round(grouped_variance, 2))
print("Grouped Std. Deviation  :", round(grouped_std, 2), "km")
print("Grouped Q1              :", round(Q1, 2), "km")
print("Grouped Q3              :", round(Q3, 2), "km")
print("Grouped IQR             :", round(grouped_IQR, 2), "km")

```
